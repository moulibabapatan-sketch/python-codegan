```python
white space or pattern removing
*strip()-it removes the white space to both side of string
*lstrip()-it remove the left side space to the string
*rstrip()-it remove the right side space to the string
```

```python
s="   mouli    "
print(s)
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s="@@@mouli@@@@"
print(s)
print(s.strip("@"))
print(s.lstrip("@"))
print(s.rstrip("@"))
```

```python
## string checking methods
islower()-it returns true if entire string is in lower case
isupper()-it returns true if entire string is in upper case
istitle()-it returns true if entire string is in title casse 
iscapitation()-it returns true if entire string is in capitalize
isdigit()-it returns true if entire string contains all digits
isalpha()-it returns true if entire string contains alphbets
isalnum()-it returns true if entire string contains either alphbets or digits
startwith(sub_string)-it returns true if string starts with substring
endwith(sub_string)-it returns true if string end with substring
```

```python
print("mouli".islower())
```
